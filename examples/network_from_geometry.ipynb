```python
from nwtrace import *
import pandas as pd
import geopandas as gpd
from pathlib import Path
import os
```

```python
lines = Path('data/baseline/vaughan/YR_lines_merged.gpkg')
nodes = Path('data/baseline/vaughan/YR_nodes_merged.gpkg')

lines_gdf = gpd.read_file(lines)
nodes_gdf = gpd.read_file(nodes).rename(columns={'FACILITYID': 'node_id'})

nodes_gdf = nodes_gdf.to_crs(lines_gdf.crs)
```

```python
lines_gdf
```

```python
network = utils.network_from_geometry(
    lines_gdf,
    nodes_gdf,
    'FACILITYID',
    'node_id',
    distance_threshold=1
)
```

```python
network.rename(columns={"from": "FROM", 'to': "TO"})[['FACILITYID', "FROM", "TO"]].to_csv('out/YR_connected')
```
